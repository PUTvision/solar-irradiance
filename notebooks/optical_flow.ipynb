````python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from skimage.registration import optical_flow_tvl1, optical_flow_ilk
````

````python
img_path_1 = '../data/Folsom/2014/09/29/20140929_002700.jpg'
img_path_2 = '../data/Folsom/2014/09/29/20140929_002900.jpg'

img1 = cv2.cvtColor(cv2.imread(img_path_1), cv2.COLOR_BGR2RGB)
img2 = cv2.cvtColor(cv2.imread(img_path_2), cv2.COLOR_BGR2RGB)

img1_grey = cv2.cvtColor(img1, cv2.COLOR_RGB2GRAY)
img2_grey = cv2.cvtColor(img2, cv2.COLOR_RGB2GRAY)
````

````python
fig, ax = plt.subplots(1, 2, figsize=(8, 4))

ax[0].imshow(img1)
ax[1].imshow(img2)
````

### Lucas-Kanade (iLK) solver

````python
flow = optical_flow_ilk(moving_image=img2_grey, reference_image=img1_grey)

fig, ax = plt.subplots(1, 4, figsize=(16, 4))

ax[0].imshow(img1)
ax[1].imshow(flow[0], cmap='bwr')
ax[2].imshow(flow[1], cmap='bwr')
ax[3].imshow(img2)
````

### TV-L1 solver

````python
flow = optical_flow_tvl1(moving_image=img2_grey, reference_image=img1_grey)

fig, ax = plt.subplots(1, 4, figsize=(16, 4))

ax[0].imshow(img1)
ax[1].imshow(flow[0], cmap='bwr')
ax[2].imshow(flow[1], cmap='bwr')
ax[3].imshow(img2)
````

### Farneback Optical Flow

````python
flow = cv2.calcOpticalFlowFarneback(img1_grey, img2_grey, None, 0.5, 3, 15, 3, 5, 1.2, 0)

# farneback_of = cv2.optflow.createOptFlow_Farneback()
# flow = farneback_of.calc(img1_grey, img2_grey, None)

fig, ax = plt.subplots(1, 4, figsize=(16, 4))

ax[0].imshow(img1)
ax[1].imshow(flow[..., 0], cmap='bwr')
ax[2].imshow(flow[..., 1], cmap='bwr')
ax[3].imshow(img2)
````

### DIS Optical Flow

````python
dis_of = cv2.DISOpticalFlow_create(preset=cv2.DISOPTICAL_FLOW_PRESET_FAST)

flow = dis_of.calc(img1_grey, img2_grey, None)

fig, ax = plt.subplots(1, 4, figsize=(16, 4))

ax[0].imshow(img1)
ax[1].imshow(flow[..., 0], cmap='bwr')
ax[2].imshow(flow[..., 1], cmap='bwr')
ax[3].imshow(img2)
````

```shell

```

````python
print(f'CUDA enabled devices: {cv2.cuda.getCudaEnabledDeviceCount()}')

gpu_prev = cv2.cuda_GpuMat()
gpu_prev.upload((img1_grey  / 255.).astype(np.float32))

gpu_frame = cv2.cuda_GpuMat()
gpu_frame.upload((img2_grey  / 255.).astype(np.float32))

print(f'Empty frames? : {gpu_frame.empty()}, {gpu_prev.empty()}')
````

### Brox Optical Flow

````python
brox_of = cv2.cuda_BroxOpticalFlow.create(
    alpha=0.197, gamma=50.0, scale_factor=0.8, inner_iterations=10, outer_iterations=77, solver_iterations=10
)

gpu_flow = brox_of.calc(gpu_prev, gpu_frame, None)

gpu_flow_x = cv2.cuda_GpuMat(gpu_flow.size(), cv2.CV_32FC1)
gpu_flow_y = cv2.cuda_GpuMat(gpu_flow.size(), cv2.CV_32FC1)
cv2.cuda.split(gpu_flow, [gpu_flow_x, gpu_flow_y], cv2.cuda.Stream_Null())

flow_x = gpu_flow_x.download()
flow_y = gpu_flow_y.download()

gpu_magnitude, gpu_angle = cv2.cuda.cartToPolar(gpu_flow_x, gpu_flow_y, angleInDegrees=False)
gpu_v = cv2.cuda.normalize(gpu_magnitude, 0.0, 255.0, cv2.NORM_MINMAX, -1)

res = gpu_v.download()

fig, ax = plt.subplots(1, 5, figsize=(20, 4))

ax[0].imshow(img1)
ax[1].imshow(flow_x, cmap='bwr')
ax[2].imshow(flow_y, cmap='bwr')
ax[3].imshow(res, cmap='bwr')
ax[4].imshow(img2)
````
